```python
import pyarrow.parquet as pq

table = pq.read_table(r"D:\2025-15\Depliegue_Soluciones_Analiticas\Proyecto\Proyecto_DSA\outputs_fase2\df_limpio.parquet")
```

```python
import pyarrow as pa
import pyarrow.compute as pc
```

```python
schema = table.schema
cols = []

for name, field in zip(table.column_names, table.columns):
    if field.type == pa.timestamp('ns'):
        cols.append(pc.cast(field, pa.timestamp('us')))
    else:
        cols.append(field)

table_fixed = pa.Table.from_arrays(cols, names=table.column_names)
```

```python
pq.write_table(
    table_fixed,
    r"D:\2025-15\Depliegue_Soluciones_Analiticas\Proyecto\Proyecto_DSA\outputs_fase2\df_limpio_FIXED.parquet",
    coerce_timestamps="us"
)
```
