**2.1. FUNCIÓN LLAMADO DE DATOS**

```python
def leer_archivos_defunciones_csv(ruta_carpeta, año_inicio, año_fin):
    """Lee y unifica archivos CSV en un rango de años, manejando distintos formatos de codificación."""
    ruta = Path(ruta_carpeta)
    archivos = [
        arch for arch in ruta.glob("Defun*.csv")
        if any(str(año) in arch.stem for año in range(año_inicio, año_fin + 1))
    ]

    if not archivos:
        print(" No se encontraron archivos CSV en el rango indicado.")
        return pd.DataFrame()

    dfs = []
    print(f"Leyendo archivos {año_inicio}-{año_fin}:")
    for archivo in tqdm(archivos):
        cargado = False
        for encoding in ["utf-8", "latin-1", "windows-1252"]:
            try:
                df = pd.read_csv(archivo, encoding=encoding, sep=',', low_memory=False)
                dfs.append(df)
                print(f" Cargado: {archivo.name} con codificación {encoding} ({len(df):,} filas)")
                cargado = True
                break
            except Exception as e:
                continue
        if not cargado:
            print(f" Error al leer {archivo.name}: no se pudo decodificar con utf-8, latin-1 ni windows-1252")

    if dfs:
        df_total = pd.concat(dfs, ignore_index=True)
    else:
        df_total = pd.DataFrame()

    return df_total
```

**3. EJECUTAR PROCESO PRINCIPAL**


```python
# =============================================================================
#PROCESO PRINCIPAL
# =============================================================================

print("=" * 60)
print("ANÁLISIS EXPLORATORIO DE DATOS - DEFUNCIONES 2021")
print("=" * 60)

# 2.1. Leer archivos de defunciones
print("\n1. LEYENDO ARCHIVOS DE DEFUNCIONES...")
df_defun = leer_archivos_defunciones_csv("data/raw/Muertes", 2021,2021)
print(f"   Total registros: {len(df_defun):,}")
print(f"   Columnas: {list(df_defun.columns)}")

print("=" * 60)
print("ANÁLISIS EXPLORATORIO DE DATOS - DEFUNCIONES 2022")
print("=" * 60)

# 2.1. Leer archivos de defunciones
print("\n1. LEYENDO ARCHIVOS DE DEFUNCIONES...")
df_defun_2 = leer_archivos_defunciones_csv("data/raw/Muertes", 2022,2022)
print(f"   Total registros: {len(df_defun_2):,}")
print(f"   Columnas: {list(df_defun_2.columns)}")
```

```python
import pandas as pd

# Columnas únicas en cada DataFrame
cols_unicas_df_defun = set(df_defun.columns) - set(df_defun_2.columns)
cols_unicas_df_defun_2 = set(df_defun_2.columns) - set(df_defun.columns)

# Lista para registrar diferencias
diferencias = []

# Columnas únicas
for col in cols_unicas_df_defun:
    diferencias.append({"Columna": col, "Diferencia": "Solo en df_defun"})
for col in cols_unicas_df_defun_2:
    diferencias.append({"Columna": col, "Diferencia": "Solo en df_defun_2"})

# Columnas comunes con tipos distintos
cols_comunes = set(df_defun.columns).intersection(set(df_defun_2.columns))
for col in cols_comunes:
    if df_defun[col].dtype != df_defun_2[col].dtype:
        diferencias.append({
            "Columna": col,
            "Diferencia": f"Tipo distinto: df_defun={df_defun[col].dtype}, df_defun_2={df_defun_2[col].dtype}"
        })

# Mostrar resultados de forma organizada
print("=== Comparación de DataFrames ===")
if not cols_unicas_df_defun and not cols_unicas_df_defun_2:
    print("✅ Los DataFrames tienen las mismas columnas.\n")
else:
    print("❌ Los DataFrames tienen columnas diferentes.\n")

if diferencias:
    df_diferencias = pd.DataFrame(diferencias)
    print("Detalle de diferencias:")
    print(df_diferencias.to_string(index=False))
else:
    print("✅ No hay diferencias en los tipos de columnas.")

```
