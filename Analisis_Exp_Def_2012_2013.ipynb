## 1. IMPORTAR LIBRERIAS

```python
import importlib
import subprocess
import sys

def verificar_e_instalar_librerias():
    """
    Verifica e instala las librerías necesarias si no están presentes.
    Puedes editar las versiones en el diccionario 'librerias'.
    """
    librerias = {
        "pycountry": None,   # Ejemplo: "23.12.11"
        "pandas": None,
        "numpy": None,
        "tqdm": None,
        "pathlib": "builtin",
        "warnings": "builtin"
    }

    for nombre, version in librerias.items():
        if version == "builtin":
            print(f"✅ {nombre} es parte de la librería estándar de Python.")
            continue

        try:
            importlib.import_module(nombre)
            print(f"✅ {nombre} ya está instalado.")
        except ImportError:
            print(f"⚠️ {nombre} no está instalado. Instalando...")
            paquete = f"{nombre}=={version}" if version else nombre
            subprocess.check_call([sys.executable, "-m", "pip", "install", paquete])
            print(f"✅ {nombre} instalado correctamente.\n")

    # ✅ Importar todas las librerías y exponerlas globalmente
    global pd, np, Path, tqdm, warnings, pycountry
    import pandas as pd
    import numpy as np
    from pathlib import Path
    from tqdm import tqdm
    import warnings
    import pycountry

    warnings.filterwarnings("ignore")
    print("\n📦 Todas las librerías listas e importadas correctamente.")
```

## 2. CREAR FUNCIONES

### 2.1. FUNCION LLAMADO DE DATOS

```python
def leer_archivos_defunciones(ruta_carpeta, año_inicio, año_fin, patron="Defun*"):
    """Lee y unifica archivos de defunciones en un rango de años
    
    Detecta automáticamente el tipo de archivo y separador
    """
    ruta = Path(ruta_carpeta)
    
    # Buscar archivos con extensiones comunes
    extensiones = ['.txt', '.csv', '.tsv']
    archivos = []
    
    for ext in extensiones:
        archivos.extend([arch for arch in ruta.glob(f"{patron}{ext}") 
                        if año_inicio <= int(arch.stem[-4:]) <= año_fin])
    
    dfs = []
    encodings = ['utf-8', 'latin-1', 'iso-8859-1', 'cp1252']
    
    for archivo in tqdm(archivos, desc="Leyendo archivos"):
        df = None
        extension = archivo.suffix.lower()
        
        # Determinar separador
        separadores = {
            '.txt': '\t',
            '.tsv': '\t',
            '.csv': ','
        }
        separador = separadores.get(extension, ',')
        
        for encoding in encodings:
            try:
                df = pd.read_csv(archivo, sep=separador, encoding=encoding, 
                               low_memory=False, dtype=str)
                df.columns = df.columns.str.upper()
                dfs.append(df)
                print(f"   ✓ {archivo.name} leído ({encoding})")
                break
            except UnicodeDecodeError:
                continue
            except Exception as e:
                if encoding == encodings[-1]:  # Solo mostrar en último intento
                    print(f"   ✗ Error en {archivo.name}: {e}")
                continue
        
        if df is None:
            print(f"   ✗ No se pudo leer {archivo.name}")
    
    if dfs:
        df_final = pd.concat(dfs, ignore_index=True)
        print(f"\n   📊 Total de archivos procesados: {len(dfs)}")
        return df_final
    else:
        print("\n   ⚠ No se encontraron archivos para procesar")
        return pd.DataFrame()
```
